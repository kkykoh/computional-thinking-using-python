```python
def string_func(*args):
    res = ''
    for word in args:
        if len(word) <= 5:
            new_letters = word[1:4].upper()   # word[1] word[2] word[3]
            if new_letters[0] == new_letters[1]:
                res += new_letters    # append 3 chars to res (initially '')
        else:
            res += word[-2:20]
    return res


print(len(string_func("aaxxz", "byyb", "cdefghijk")))
#                                ___           __
#                      +0        +3            +2
```

```python
for i in range(1,4):
    for j in range(1,4):
        if j % i == 1:
            break
        print(i, j, end = "-")
        
```

```python
# i = 1       i j 
#   j = 1     1 1  -- j%i = 1%1 = 0    p  --> 1 1-
#   j = 2     1 2  --       2%1 = 0    p
#   j = 3     1 3  --       3%1 = 0    p
# i = 2
#   j = 1          -- j%i   1%2 = 1    <<< break
 
 
# i = 3
#   j = 1          -- j%i   1%3 = 1    <<< break
```

```python
for i in range(1,4):
    for j in range(1,4):
        if j % i == 1:
            continue
        print(i, j, end = "-")
```

```python
# i = 1       i j
#   j = 1     1 1  -- j%i = 1%1 = 0    p
#   j = 2     1 2  --       2%1 = 0    p
#   j = 3     1 3  --       3%1 = 0    p
# i = 2
#   j = 1          -- j%i   1%2 = 1    <<< continue is trigged (therefore, skip)
#   j = 2     2 2  --       2%2 = 0    p
#   j = 3          --       3%2 = 1    <<< continue is trigged (therefore, skip)
# i = 3
#   j = 1          -- j%i   1%3 = 1    <<< continue is trigged (therefore, skip)
#   j = 2     3 2  --       2%3 = 2    p
#   j = 3     3 3  --       3%3 = 0    p
```
