### Exercise 1: Calculate BMI

Body Mass Index is a simple calculation using a person’s height and weight. The formula is $BMI = kg/m^2$ where $kg$ is a person’s weight in kilograms and $m^2$ is their height in metres squared. </br>
A BMI of 25.0 or more is overweight, while the healthy range is 18.5 to 24.9. </br>
Write a bmi calculator to understand your bmi weight status.

```python
float(input("Input your dream weight:"))
```

```python
weight = float(input("Input your dream weight:"))
height = float(input("Input your dream height:"))

bmi = weight/(height)**2

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Healthy")
else:
    print("Overweight")
```

### Exercise 2: Calculate Carbon

When you execute a heavy computational workload in the cloud, you may notice that the final environmental impact exceeds the raw energy used by the CPU. This is primarily due to two environmental multipliers: Grid Carbon Intensity and the data center's Power Usage Effectiveness (PUE). While the server's base energy is simply its hardware power (kW) multiplied by its runtime (hours), the PUE typically adds a 20% overhead for facility cooling. Some unoptimized codebases even impose an "inefficient algorithm penalty" that drastically increases the total runtime. 

There are two types of cloud regions you can deploy to: 
1. Clean Region 
2. Standard Region 

You also need to speficy if the "Inefficient Algorithm Penalty" is to be applied.

Write a carbon footprint calculator for estimating the actual emissions.


```python
power = float(input("Enter server hardware power in kW (e.g., 0.15 for 150W): "))
runtime = float(input("Enter base script runtime in hours (e.g., 1.5): "))
region = input("Select region type (1 or 2): ")
inefficient_surcharge = input("Apply Inefficient Algorithm Penalty? (Y/N): ")

pue = 0.2          # Power Usage Effectiveness
clean = 0.030      # kg CO2/kWh
standard = 0.450   # kg CO2/kWh
inefficient = 10

if region == "1":
    if inefficient_surcharge == "N":
        total_emissions = power * runtime * clean * (1 + pue)
    else:
        total_emissions = power * runtime * clean * (1 + pue) * inefficient
else:
    if inefficient_surcharge == "N":
        total_emissions = power * runtime * standard * (1 + pue)
    else:
        total_emissions = power * runtime * standard * (1 + pue) * inefficient
        

print("Total emissions is {:.4f} kg CO2".format(total_emissions))
```

### Exercise 3: Shorten a Word

Letters between the first and last letters of a word may be replaced by the number of letters omitted. </br>
For example, the word "internationalization" can be abbreviated by replacing the eighteen middle letters ("nternationalizatio") with "18", leaving "i18n". </br>
These word shortenings are sometimes called numerical contractions. </br>
Examples of numerical contractions include:
- g11n: globalization
- i14y – interoperability
- a11y – accessibility
- p13n – personalization
- s5n – shorten
- l10n – localization
- i18n – internationalization

Write a code to help generate the numerical contractions.

```python
word = input()

count = len(word) - 2
shorten = word[0] + str(count) + word[-1]

## whiteboard
print(word[0], str(count), word[-1])
print(shorten)
```
