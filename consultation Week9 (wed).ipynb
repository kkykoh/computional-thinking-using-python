```python
class A:
    def __init__(self):
        self.a = 1

    def sum(self, b):
        return self.a + b.a
        

class B(A):
    def __init__(self):
        super().__init__()
        self.b = 2
    
    def sum(self, b):
        return self.b + b.a


# a = A()   # a.a = 1
# b = B()   # b.a = 1, b.b = 2

# print(a.sum(b))

obj_a = A()   # obj_a.a = 1
obj_b = B()   # obj_b.a = 1, obj_b.b = 2

obj_a.sum(obj_b)   # we are running the sum method of obj_a
                   # self.a + b.a
                   # replace `self` with `obj_a`, replace `b` with `obj_b`
                   # obj_a.a + obj_b.a
```

```python
class A:
    def __init__(self):
        self.a = 1

    def sum(self, b):
        return self.a + b.a
        

class B(A):
    def __init__(self):
        super().__init__()
        self.b = 2
    
    def sum(self, b):
        return self.b + b.a


obj_a = A()
obj_b = B()

print(obj_a.sum(obj_b))


# obj_a.sum(obj_b)  ---> obj_a.a + obj_b.a

##  obj_a.sum(______) 
##  obj_a is an instance of Class A
##  go to class A, see the method name `sum`, copy over
##                                            self.a + b.a
##  replace 'self.' with `obj_a.`, and 'b.' with `obj_b.`
##  obj_a.a + obj_b.a
```

```python
import re

text = "how are you good day mate lion tiger zebra"
# p = re.compile(r'[a-c-x-z]')   # re.compile(r'[abcxyz-]')   
p = re.compile(r'[d-f-p-s]')  # [defpqrs-]
m = p.findall(text)
print(m)

```

```python
p = re.compile(r'\w.at', re.M)   # \w [a-zA-Z0-9_]
m = p.findall('What bhat chat is this about 5@at 6#at _7at?')
print(m)
print(len(m))
```
