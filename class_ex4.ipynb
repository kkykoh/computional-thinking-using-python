```python
def simple(month):
    if month == 2:
        return 28
    elif month <= 7 == month % 2:
        return 31
    else:
        return 30

simple(2)
```

```python
def is_leap_year(year:int):
    if year % 4 != 0:
        return False
    elif year % 100 != 0:
        return True
    elif year % 400 != 0:  # years like 1800, 1900, 2100 are not leap years although they are divisible by 4
        return False
    else:                  # years like 1600, 2000, 2400 are.
        return True
```

```python
def last_day_of_month(year:int, month:int):
    if month <= 0 or month > 12:
        print('not a valid month')
        return -1
    if month == 2:
        if is_leap_year(year):
            return 29
        else:
            return 28
    elif (month <= 7) == (month % 2):    # what if parentheses are removed?
        return 31
    else:
        return 30
```

```python
print(last_day_of_month(2024, 2), last_day_of_month(2100, 2))
```

### Can "is_leap_year" be inside "last_day_of_month" ?

```python
def last_day_of_month(year:int, month:int):
    if month <= 0 or month > 12:
        print('not a valid month')
        return -1
    if month == 2:
        if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
            return 29
        else:
            return 28
    elif (month <= 7) == (month % 2):
        return 31
    else:
        return 30
```

### Exercise 4-1: Ambiguity with i18n

```python
def shorten(s):
    if len(s) <= 6:
        return s
    else:
        return s[0] + str(len(s)-2) + s[-1]

d = {}
with open('week4_ex1.txt', 'r') as f:
    for s in f:
        s = s.strip().lower()
        acron = shorten(s)
        if acron in d:
            d[acron].append(s)
        else:
            d[acron] = [s]

print('There exists ambiguity when shortening the words below:')
for k, v in d.items():
    if len(v) > 1:
        print(k, v)
```

### Exercise 4-2: Personal Income Tax

```python
def cal_income_tax (personal_income):
    tax_bracket = [
        [20000, 0], [10000, 2], [10000, 3.5], [40000, 7], [40000, 11.5], [40000, 15],
        [40000, 18], [40000, 19], [40000, 19.5], [40000, 20], [180000, 22], [500000, 23]]
    remaining_rate = 24

    remaining_income = personal_income
    tax_amt = 0
    for amt, rate in tax_bracket:
        if amt <= remaining_income: # use the full amount
            tax_amt += amt * rate / 100
            remaining_income = remaining_income - amt
        else:
            tax_amt += remaining_income * rate / 100
            remaining_income = 0
            break
    if remaining_income > 0:
        tax_amt += remaining_income * remaining_rate / 100
    return tax_amt
```

```python
import random, time             # will explain import in details at week 6
random.seed(time.time())
group_income = []
for i in range(100):
    group_income.append(random.gauss(60000, 10000))
# why not using group_income = [random.gauss(60000, 10000)] * 100
# print(group_income)
group_tax = []
for i in range(100):
    group_tax.append(cal_income_tax(group_income[i]))
# print(group_tax)

def cal_mean_std(tax_list):
    avg_tax = sum(tax_list) / len(tax_list)
    sum_sq = 0.0
    for tax in tax_list:
        sum_sq += (tax - avg_tax) ** 2
    return avg_tax, (sum_sq / len(tax_list)) ** 0.5

tax_mean, tax_std = cal_mean_std(group_tax)
print('the mean and standard deviation of the personal taxes from this group: {:.2f}, {:.2f}'.format(tax_mean, tax_std))
```

```python
def check_dimensionality(dim, p1, p2):
    if len(p1) != len(p2) or (dim >= 0 and len(p1) != dim):
        return -1            # return -1 if dimensionality unmatched
    if dim == -1:
        dim = len(p1)
    return dim

def Euclid_dist(*args, **kwargs):
    dim, args_bgn = -1, 0                # dim starts with undefined
    if 'dim' in kwargs:
        dim = kwargs['dim']
    elif len(args) % 2 == 1:
        dim = args[0]
        args_bgn = 1
    if 'P1' in kwargs and 'P2' in kwargs:
        p1, p2 = kwargs['P1'], kwargs['P2']
        dim = check_dimensionality(dim, p1, p2)
        if dim == -1:
            return 'dimensionality unmatched'
    elif len(args) == args_bgn + 2 and type(args[args_bgn]) == list and type(args[args_bgn+1]) == list:
        p1, p2 = args[args_bgn], args[args_bgn+1]
        dim = check_dimensionality(dim, p1, p2)
        if dim == -1:
            return 'dimensionality unmatched'
    elif dim >= 0 and len(args) != args_bgn + 2 * dim:
        return 'dimensionality unmatched'
    else:
        if dim == -1:
            dim = (len(args) - args_bgn) // 2
        p1, p2 = args[args_bgn:args_bgn+dim], args[args_bgn+dim:args_bgn+2*dim]
    sum_sq = 0.0
    for i in range(dim):
        sum_sq += (p1[i] - p2[i]) ** 2
    return sum_sq ** 0.5
```

```python
def calculate(p1, p2):  # now implement for two lists p1, p2 representing the lists of coordinates
    sum_sq = 0.0
    for i in range(len(p1)):
        sum_sq += (p1[i] - p2[i]) ** 2
    return sum_sq ** 0.5
```

```python
def Euclid_dist(*args, **kwargs):

    # there are 7 use cases in the cell below
    # how to write one function to cater for all cases?
    if len(kwargs) == 0:
        if len(args) % 2 == 0:
            # Case 1
            return calculate(args[:len(args)//2], args[len(args)//2:])
        elif len(args) == (1 + 2 * args[0]):
            # Case 2
            return calculate(args[1:len(args)//2+1], args[len(args)//2+1:])
        else:
            # Case 3
            return calculate(args[1], args[2])
    else:
        if "dim" not in kwargs:
            # Case 7
            return calculate(kwargs["P1"], kwargs["P2"])
        elif len(args) == 0:
            # Case 6
            return calculate(kwargs["P1"], kwargs["P2"])
        elif len(args) == (kwargs["dim"] * 2):
            # Case 4
            return calculate(args[:len(args)//2], args[len(args)//2:])
        else:
            # Case 5
            return calculate(args[0], args[1])
```

### Exercise 4-4: Affordability

My yearly salary increases by 500, the price of chicken rice increases by 5% every year on average, how many plates of chicken rice can i afford 20 years down the road?

```python
salary_first_year, chk_rice_first_year = 70000, 3.5
diff, ratio = 500, 1.05
affordability = ['year '+ str(int(time.ctime()[-4:]) + i) + ': ' + str(int((salary_first_year + diff * i) / (chk_rice_first_year * ratio ** i))) + ' plates of chicken rice' for i in range(20)]
print('\n'.join(affordability))
```
