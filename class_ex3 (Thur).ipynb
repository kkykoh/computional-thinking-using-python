```python
brightest_color, brightest_value = '', 0.0
with open('week3_ex3.txt', 'r') as f:
    lines = f.readlines()
    for line in lines:
        tokens = line.strip().split('\t')
        r, g, b = int(tokens[1][1:3], 16) / 255, int(tokens[1][3:5], 16) / 255, int(tokens[1][5:7], 16) / 255
        rel_lumin = 0.2126 * r + 0.7152 * g + 0.0722 * b
        if rel_lumin > brightest_value:
            brightest_color, brightest_value = tokens[0], rel_lumin

print('the brightest color is %s, with relative luminance %.4f' % (brightest_color, brightest_value))
```

### Exercise 3-5: maximal palindromes

```python
with open('week3_ex5.txt', 'r') as f:
    lines = f.readlines()
    for line in lines:
        s = line.strip()
        s_len, curr_max, max_palind = len(s), 0, ''
        # case 1: palindromes with odd length
        for ctr_pos in range(0, s_len):                                     # any character can be a center of a palindrome
            palind_len, palind_bgn, palind_end = 1, ctr_pos, ctr_pos+1      # s[palind_bgn:palind_end] is a palindrome
                                                                            # centered at ctr_pos, try make this range larger
            while palind_bgn >= 1 and palind_end < s_len and s[palind_bgn-1] == s[palind_end]:
                palind_len += 2
                palind_bgn -= 1
                palind_end += 1
            if palind_len > curr_max:                                       # if ever get a longer palindrome
                curr_max, max_palind = palind_len, s[palind_bgn:palind_end]
                
        # case 2: palindromes with even length
        for ctr_pos in range(1, s_len):                                     # consider a center between ctr_pos-1 and ctr_pos 
            palind_len, palind_bgn, palind_end = 0, ctr_pos, ctr_pos        # s[palind_bgn:palind_end] is a palindrome
                                                                            # centered at ctr_pos, try make this range larger
            while palind_bgn >= 1 and palind_end < s_len and s[palind_bgn-1] == s[palind_end]:
                palind_len += 2
                palind_bgn -= 1
                palind_end += 1
            if palind_len > curr_max:                                       # if ever get a longer palindrome
                curr_max, max_palind = palind_len, s[palind_bgn:palind_end]
        print('the longest palindrome is "' + max_palind + '" with length', curr_max)
```

### Exercise 3-6: circle packing in rectangles

```python
import math

s = input('enter p, q as the side lengths in meters of a rectangle: ')
if ',' in s:
    tokens = s.split(',')
else:
    tokens = s.split()
p, q = int(tokens[0]), int(tokens[1])
m = p * q
level = 3
while True:
    height = (level - 1) / 2 * math.sqrt(3) + 1     # height at the zig-zag part
    if height > p and height > q:
        break
    if height <= q:     # taking side with length p as base
        m = max(m, p * level - level // 2 + int(q - height) * p)
    if height <= p:     # taking side with length q as base
        m = max(m, q * level - level // 2 + int(p - height) * q)
    level += 2
print('maximum number of circles: ', m)
```

### Exercise 3-7: recurring transfers

```python
bgn = input('enter the first month to transfer in the format of mm/yyyy: ')
tokens = bgn.split('/')
bgn_year, bgn_month = int(tokens[1]), int(tokens[0])
end = input('enter the last month to transfer in the format of mm/yyyy: ')
tokens = end.split('/')
end_year, end_month = int(tokens[1]), int(tokens[0])
curr_year, curr_month = bgn_year, bgn_month
while curr_year < end_year or (curr_year == end_year and curr_month <= end_month):
                    # can the above be changed to "while curr_year <= end_year and curr_month <= end_month:"?
    if curr_month == 2:
        last_day = 28 if curr_year % 4 != 0 or (curr_year % 100 == 0 and curr_year % 400 != 0) else 29
    elif curr_month in (1, 3, 5, 7, 8, 10, 12):
        last_day = 31
    else:
        last_day = 30
    print('{:02d}/{:02d}/{:04d}'.format(last_day, curr_month, curr_year))
    curr_month += 1
    if curr_month == 13:
        curr_year += 1
        curr_month = 1
```

### Exercise 3-8: finding primes

```python
# set upper_bound here, you may change it
upper_bound = 200
```

```python
prime = []
for i in range(2, upper_bound+1):        # test all integers from 2 to upper_bound
    is_prime = True                      # first assume i is a prime
    for f in range(2, i):                # for each integer f from 2 to i-1, check if f is a divisor of i
        if i % f == 0:
            is_prime = False             # if divisible, set is_prime to False if divisible
    if is_prime:
        prime.append(i)
print(prime)
```
