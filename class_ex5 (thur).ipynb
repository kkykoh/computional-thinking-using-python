```python
def sqrtBS_for16(x): 
    minx, maxx = 0, x
    mid = (minx + maxx) / 2
    while mid ** 2 != x:  ## on the basis that int sol is guaranteed
        if mid ** 2 > x: 
            maxx = mid
        else: 
            minx = mid
        mid = (minx + maxx) / 2
    return mid
```

```python
sqrtBS_for16(16)
```

```python
def sqrtBS(x, e = 1e-3): 
    minx, maxx = 0, x
    mid = (minx + maxx) / 2
    while abs(mid ** 2 - x) > e:  ## continue for as long as error is big
        print("candidate is", mid)
        if mid ** 2 > x: 
            maxx = mid
        else: 
            minx = mid
        mid = (minx + maxx) / 2
    return mid
```

```python
def nrtBS(x, n, e = 1e-3):
    minx, maxx = 0, x
    mid = (minx + maxx) / 2
    while abs(mid ** n - x) > e:
        if mid ** n > x: 
            maxx = mid
        else: 
            minx = mid
        mid = (minx + maxx) / 2
    return mid
```

```python
nrtBS(16, 2)
```

```python
nrtBS(10, 2)
```

```python
def LinearSearch(ls, target):
    for i in range(len(ls)):
        if ls[i] == target:
            return i
    return -1
```

### Exercise 5-4: Flash Flooded Skyline

```python
height = [4, 18, 5, 0, 10, 10, 60, 40, 20, 30, 28, 36]
rainfall = [20, 30, 50]
```

```python
sky0 = [(1.0 * height[i], 20 * i, 20 * (i + 1)) for i in range(len(height))]
print(sky0)
```

```python
def merge(skyline):
    if len(skyline) == 1:
        return skyline
    skyline_new = []
    curr_h, curr_l, curr_r = skyline[0][0], skyline[0][1], skyline[0][2]
    for seg in skyline[1:]:
        if seg[0] == curr_h:
            curr_r = seg[2]    # push the right boundary of (curr_h, curr_l, curr_r) to the right side of this segment
        else:
            skyline_new.append((curr_h, curr_l, curr_r))    # write to the new skyline, and start a new
            curr_h, curr_l, curr_r = seg[0], seg[1], seg[2]
    skyline_new.append((curr_h, curr_l, curr_r))
    return skyline_new
```

```python
sky1 = merge(sky0)
print(sky1)
```

```python
def flood(skyline, fall):
    if len(skyline) == 1:    # only one segment, height will increase by 'fall' over the entire interval
        return [(skyline[0][0] + fall, skyline[0][1], skyline[0][2])], 0
    peak, valley = [0], []    # the left wall is a peak
                              # peak: a list of locations
                              # valley: a list of (index_in_skyline, how_much_must_it_grow_to_level_with_an_adjacent_segment)
    if skyline[0][0] < skyline[1][0]:    # if the first segment is a valley, otherwise ignore
        valley.append((0, skyline[1][0] - skyline[0][0]))
    for i in range(1, len(skyline) - 1):
        if skyline[i][0] > skyline[i-1][0] and skyline[i][0] > skyline[i+1][0]:    # taller than both sides, it is a peak
            peak.append((skyline[i][1] + skyline[i][2]) // 2)
        if skyline[i][0] < skyline[i-1][0] and skyline[i][0] < skyline[i+1][0]:    # shorter than both sides, it is a valley
            valley.append((i, min(skyline[i-1][0] - skyline[i][0], skyline[i+1][0] - skyline[i][0])))
    peak.append(skyline[-1][2])    # the right wall is a peak
    if skyline[-1][0] < skyline[-2][0]:    # if the last segment is a valley, otherwise ignore
        valley.append((len(skyline)-1, skyline[-2][0] - skyline[-1][0]))
    # all peaks and valley are identified up to this point
    # for each valley, need to consider what is the amount of rainfall to make it level with an adjacent segment
    min_fall = fall
    for i in range(len(valley)):
        j = valley[i][0]    # i is the index on valley, j is the index on skyline, valley[i] is between peak[i] and peak[i+1]
        min_fall = min(min_fall, (skyline[j][2] - skyline[j][1]) * valley[i][1] / (peak[i+1] - peak[i]))
        # for the current valley, the amount of water needed to level with an adjacent segment is
        # width * height where width = (skyline[j][2] - skyline[j][1]) and height = valley[i][1]
        # but it collects all water between peak[i] and peak[i+1]
    for i in range(len(valley)):
        j = valley[i][0]
        skyline[j] = (skyline[j][0] + (peak[i+1] - peak[i]) * min_fall / (skyline[j][2] - skyline[j][1]), skyline[j][1], skyline[j][2])
        # the increment amount is the min_fall * (peak[i+1] - peak[i]), reshape into width (skyline[j][2] - skyline[j][1])
        # the increment height on this particular valley is (peak[i+1] - peak[i]) * min_fall / (skyline[j][2] - skyline[j][1])
    return merge(skyline), fall - min_fall
    # since there is valley leveled with an adjacent segment, skyline has to be merged.
```

```python
for fall in rainfall:
    skyline = sky1.copy()
    while fall > 0:
        skyline, fall = flood(skyline, fall)
        # each time, fall is the remaining rainfall amount after leveling at least one valley in the skyline
        # skyline are updated at the same time by flood function
    print('; '.join(['%.3f, %d' % (t[0], t[2] - t[1]) for t in skyline]))
```
