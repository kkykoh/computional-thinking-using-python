### Exercise 1

In a typical password checker, there are a set of conditions which are often checked to ensure that the password meets the required complexity of the platform. In this exercise, your job is to check whether a given password satisfies the below list of conditions:

- The password should be between length 8 to 15 (inclusive)
- The password should contain one of each of the following
    - An upper-case letter
    - A lower-case letter
    - A digit
    - A special character

````python
password = input() # "C0mputation@l"

### insert solution here
# we can decompose the problem into multiple subproblems depending on the different conditions
# brute force solution: a simple way to check is to consider each subproblem and code separately
valid = True
# check for length
if 8 <= len(password) and len(password) <= 15:

    # case 1: checking for upper-case letter
    found = False
    for i in range(len(password)):
        if password[i].isupper():
            found = True

    if not found:
        valid = False
    
    # case 2: checking for lower-case letter
    found = False
    for i in range(len(password)):
        if password[i].islower():
            found = True

    if not found:
        valid = False

    # case 3: checking for a digit
    found = False
    for i in range(len(password)):
        if password[i].isdigit():
            found = True

    if not found:
        valid = False

    # case 4: checking for a special character
    found = False
    for i in range(len(password)):
        if (not password[i].isdigit()) and (not password[i].islower()) and (not password[i].isupper()):
            found = True

    if not found:
        valid = False
    
else:
    valid = False

if valid:
    print("Valid")
else:
    print("Invalid")
        
````

````python
# using ls[start:] or ls[start:large_no] takes all elements starting from the start index
print(ls[1:])
print(ls[1:999])
````

````python
# using ls[:end] or ls[0:end] takes all elements up till (and not including) the end index
print(ls[:4])
print(ls[0:4])
````

````python
# you can also use negative indexes
print(ls[1:4])
print(ls[-4:-1]) 
print(ls[1:-1]) 
````

````python
# using ls[start:end:step] takes a step size to slice elements of the list
print(ls[1:4:2])

# you can also use negative step sizes
print(ls[3:0:-2])
````

### for Loops with iterables

````python
# using lists
ls = [1, 2, 3, 4, 5]

# method 1: directly using the list
for i in ls:
    print(i)

# method 2: using range and len
for i in range(len(ls)):
    print(ls[i])
````

````python
# using dictionaries
d = {'deer': 8, 'goat': 11, 'wolf': 15}

# method 1: directly using the dictionary
for i in d: # this loops through the keys
    print(d[i])

# method 2: using items
for k, v in d.items(): # k gets assigned the key, and v gets assigned the value
    print(k, v)
````

### Exercise 2

The calculation of income tax rates depends on an individual's tax residency status. Singapore's personal income tax rates for resident taxpayers are progressive. This means that higher income earners pay a proportionately higher tax, with the current highest personal income tax rate at 24%.

Write a Python code to compute the resident tax given a certain assessible income.

Source: https://www.iras.gov.sg/taxes/individual-income-tax/basics-of-individual-income-tax/tax-residency-and-tax-rates/individual-income-tax-rates 

In this part, we can explore different ways / tools to store the resident tax data, and observe how the solution changes.

````python
# using a list/tuple, can store the format in the following form
data = [
    [20000, 0],
    [10000, 2],
    [10000, 3.5],
    [40000, 7],
    [40000, 11.5],
    [40000, 15],
    [40000, 18],
    [40000, 19],
    [40000, 19.5],
    [40000, 20],
    [180000, 22],
    [500000, 23]
]

remaining = 24

# obtain the income you want to assess
income = 70000

remaining_income = income
tax_amt = 0
# use a for loop to loop through all items in data, until you reach the end
for amt, rate in data:
    if amt <= remaining_income: # we can use the full amount
        tax_amt = tax_amt + amt * rate / 100
        remaining_income = remaining_income - amt
    else:
        tax_amt = tax_amt + remaining_income * rate / 100
        remaining_income = 0

# settle the remaining income
if remaining_income > 0:
    tax_amt = tax_amt + remaining_income * remaining / 100

print("for an income of ${}, the person is taxed ${}".format(income, tax_amt))
````

````python
# how about a dictionary?
data = {
    "0 - 20000": 0,
    "20000 - 30000": 2,
    "30000 - 40000": 3.5,
    "40000 - 80000": 7,
    "80000 - 120000": 11.5,
    "120000 - 160000": 15,
    "160000 - 200000": 18,
    "200000 - 240000": 19,
    "240000 - 280000": 19.5,
    "280000 - 320000": 20,
    "320000 - 500000": 22,
    "500000 - 1000000": 23
}

remaining = 24

# obtain the income you want to assess
income = 70000

tax_amt = 0
# use a for loop to loop through all items in data, until you reach the end
for range_values, rate in data.items():
    low = int(range_values.split(" - ")[0])
    high = int(range_values.split(" - ")[1])

    # now we can use decomposition
    # case 1: income more than high
    if income > high:
        tax_amt = tax_amt + (high - low) * rate / 100
    # case 2: income between low and high
    elif low <= income and income <= high:
        tax_amt = tax_amt + (income - low) * rate / 100
    # case 3: income less than low
    else:
        tax_amt = tax_amt + 0

# settle the remaining income
if income > 1000000:
    tax_amt = tax_amt + (income - 1000000) * remaining / 100

print("for an income of ${}, the person is taxed ${}".format(income, tax_amt))
````

### Exercise 3

A rectangle in 2D can be represented by their $x$ and $y$ coordinates of two diagonally opposite points (usually the top left and bottom right corners).

In this problem, you are given two rectangles (their top left and bottom right coordinates). Your objective is to calculate whether the two rectangles intersect (i.e. overlap each other)

````python
points_a = ((3, 6), (5, 4))
points_b = ((4, 7), (6, 5))

## using cases
# for two rectangles to NOT intersect
# 1. right line of a < left line of b
# 2. right line of b < left line of a
# 3. top line of a < bottom line of b
# 4. top line of b < bottom line of a

if points_a[1][0] < points_b[0][0]:
    print("not intersecting")
elif points_b[1][0] < points_a[0][0]:
    print("not intersecting")
elif points_a[0][1] < points_b[1][1]:
    print("not intersecting")
elif points_b[0][1] < points_a[1][1]:
    print("not intersecting")
else:
    print("intersecting")
````

````python
points_a = ((3, 6), (5, 4))
points_b = ((4, 7), (6, 5))

## another way
# in 1D, two lines intersect by checking the condition max(line1_left, line2_left) < min(line1_right, line2_right)
# in 2D, we will need to check two set of 1D lines
left = max(points_a[0][0], points_b[0][0])
right = min(points_a[1][0], points_b[1][0])
bottom = max(points_a[1][1], points_b[1][1])
top = min(points_a[0][1], points_b[0][1])

if left < right and bottom < top:
    print("intersecting")
else:
    print("not intersecting")
````
