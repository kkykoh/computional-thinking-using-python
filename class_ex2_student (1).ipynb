### Range and for loops

```range``` conveniently constructs an arithmetic series.

### Exercise 1

In a typical password checker, there are a set of conditions which are often checked to ensure that the password meets the required complexity of the platform. In this exercise, your job is to check whether a given password satisfies the below list of conditions:

- The password should be between length 8 to 15 (inclusive)
- The password should contain one of each of the following
    - An upper-case letter
    - A lower-case letter
    - A digit
    - A special character

````python
'a'.isupper()
'A'.isupper()
````

````python
found = True # will not print because you set true

if not found:
    print("this")
````

````python
found = False 

if not found: # if not true, will print
    print("that") # prints that because it is true
````

````python
password = input() # "C0mputation@l"

### insert solution here

if len(password) < 8 and len(password) > 15:

# case 1: is there at least one upper case
    found = False

    for i in range(len(password)):
        if password[i].isupper():
            found = True

    # this will happen aft checking every letter
    # and none is an Uppercase letter
    if not found:
        valid = False

    # valid = True --> if this is still true here, then it passed case 1
    # valid = False --> if this is false, then it failed case 1
    else: 
        # case 2: is there at least one lower case
        found = False

        for i in range(len(password)):
            if password[i].islower():
                found = True

        # this will happen aft checking every letter
        # and none is an lowercase letter
        if not found:
            valid = False

        # valid = True --> if this is still true here, then it passed both case 1 & 2
        # valid = False --> if this is false, then it failed 1 or both cases

        else:
            # case 3: got digits?
            found = False

            for i in range(len(password)):
                if password[i].isdigit():
                    found = True

            # this will happen aft checking every letter
            # and none is an number
            if not found:
                valid = False

            else:
                # case 4: got special characters
                found = False

                for i in range(len(password)):
                    # in this case, special char is something that is not lowercase, not uppercase and not digits
                    if (not password[i].isdigit()) and (not password[i].islower()) and (not password[i].isupper):
                        found = True

    # this will happen aft checking every letter
    # and none is an number
    if not found:
        valid = False

else: 
    valid = False

if valid:
    print("valid")

else: 
    print("not valid")
````

Indexing a list

````python
ls = [5, 6, 7, 8, 9]
print(len(ls)) # len() gives you the length of a list
````

````python
# we can slice one element using ls[index]
# index starts from 0, and last index is len(ls)-1
print(ls[2])
````

````python
# we can slice a subset of the list using ls[start:stop:step] (note how this is similar to range)
print(ls[1:4])
````

````python
# using ls[start:] or ls[start:large_no] takes all elements starting from the start index
print(ls[1:])
print(ls[1:999])
````

````python
# using ls[:end] or ls[0:end] takes all elements up till (and not including) the end index
print(ls[:4])
print(ls[0:4])
````

````python
# you can also use negative indexes
print(ls[1:4])
print(ls[-4:-1]) 
print(ls[1:-1]) 
````

````python
# using ls[start:end:step] takes a step size to slice elements of the list
print(ls[1:4:2])

# you can also use negative step sizes
print(ls[3:0:-2])
````

````python
print(ls[0:3:-2]) 

# since there's nothing, it will not give error
# it just returns blank
````

### for Loops with iterables

````python
# using lists
ls = [1, 2, 3, 4, 5]

# method 1: directly using the list
for i in ls:
    print(i)

# method 2: using range and len
for i in range(len(ls)):
    print(ls[i])
````

````python
# using dictionaries
d = {'deer': 8, 'goat': 11, 'wolf': 15}

# method 1: directly using the dictionary
for i in d: # this loops through the keys
    print(d[i])

# method 2: using items
for k, v in d.items(): # k gets assigned the key, and v gets assigned the value
    print(k, v)
````

````python
# using strings
s = 'abcdef'

# method 1: directly using the string
for a in s:
    print(a)

# method 2: using range and len
for i in range(len(s)):
    print(s[i])
````
