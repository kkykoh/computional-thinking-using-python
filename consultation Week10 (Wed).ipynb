```python
holder = ["rush_hour_iii", "mission", "impossible", 'njl+ki', "jamesbond"]

for text in holder:
    p = re.compile(r'[i-n]{2,}')   # [ijklmn]
    m = p.findall(text)
    print(m)
```

```python
text = "Let's look at more examples dealing with character class"

p = re.compile(r'[a-c-g-i]{2}')  # r'[abcghi-][abcghi-]'
m = p.findall(text)
print(m)
```

```python
text = "hello goodbye"
p = re.compile(r'((\w){3})')
m = p.findall(text)
print(m)
```

```python
p = re.compile(r'((\w){3,})')
m = p.findall('abcdefg hijklmn')
print(m)
```

```python
p = re.compile(r'look|at')
m = p.findall('looker cat')
print(m)
```

```python
text = 'Before you say helpme, ask if you had already done everything uu could to help yourself'

p1 = re.compile(r'(([aeiou])\w+\2)')      # one vowel + 1 or more \w + that same vowel 
m1 = p1.findall(text)
print('m1: ', m1)

p2 = re.compile(r'(([aeiou])\w*\2)')      # one vowel + any number of \w + that same vowel
m2 = p2.findall(text)
print('m2: ', m2)

p3 = re.compile(r'(([aeiou])\w?\2)')      # one vowel + 0 or 1 \w + that same vowel
m3 = p3.findall(text)
print('m3: ', m3)
```
