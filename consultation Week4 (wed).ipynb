```python
ls = [601, 602, 603, 604, 605]

task_completed = []

count = 0
while count < len(ls):
    print("Currently, ls", ls)
    task_completed.append(
        ls.pop(0)
    )
    count += 1
    print("  count: ", count, "task_completed", task_completed)

print(count)
print(task_completed)
```
