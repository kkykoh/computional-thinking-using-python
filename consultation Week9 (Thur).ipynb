```python
class A:
    def __init__(self):
        self.a = 1

    def sum(self, b):
        return self.a + b.a
        

class B(A):
    def __init__(self):
        super().__init__()
        self.b = 2
    
    def sum(self, b):
        return self.b + b.a


a = A()
b = B()

print(a.sum(b))
```

```python
class A:
    def __init__(self):
        self.a = 1

    def sum(self, b):
        return self.a + b.a
        

class B(A):
    def __init__(self):
        super().__init__()
        self.b = 2
    
    def sum(self, b):
        return self.b + b.a


obj_a = A()     # obj_a.a = 1
obj_b = B()     # obj_b.a = 1, obj_b.b = 2

print(obj_a.sum(obj_b))


# obj_a.sum(obj_b)  ---> obj_a.a + obj_b.a

##  obj_a.sum(______) 
##  obj_a is an instance of Class A
##  go to class A, see the method name `sum`, copy over
##                                            self.a + b.a   ### from method of Class A
##  replace the objects on the left
##  self refers to obj_a, b refers to obj_b 
##  obj_a.a + obj_b.a
```
