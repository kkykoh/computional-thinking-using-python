```python
class Course():

    def __init__(self, name, term, special=False):
        self.name = name
        self.term = term 
        self.special = special
    
    def set_as_special(self):
        self.special = True

    def describe_course(self):
        if self.special:
            return "The special course {} will be in {}".format(self.name, self.term)

        else:
            return "The special course {} will be in {}".format(self.name, self.term)

course = Course('Advance coke appreciation', 'nov 2023')
course.set_as_special()

# print(course)
assert course.describe_course() == "The special course Advance coke appreciation will be in nov 2023"
```
