```python
data = [7, 12, 5, 18, 11]

def mix(a, b):
    if a > b:
        return a - b
    else:
        return b - a + 1

def scan(items, x):
    for i in range(len(items)):
        if items[i] % x == 0:
            return i
    return -1

def crunch(items):
    result = 0
    for n in items:
        if n % 2 == 0:
            result = result + n // 2
        else:
            result = result - 1
    return result



print(mix(4, 9))
print(scan(data, 6))
print(crunch(data))
print(mix(crunch(data), scan(data, 2)))

final = []
for i in range(len(data)):
    if data[i] % 9 == 0:
        final.append((i, data[i]))

print(final)
```

```python
def total_steps(steps):
    total = 0
    for s in steps:
        total += s
    return total

def average_steps(steps):
    return sum(steps) // len(steps)

def first_goal_day(steps, goal):
    for i in range(len(steps)):
        if steps[i] >= goal:
            return i
    return -1

def count_goal_days(steps, goal):
    count = 0
    for s in steps:
        if s >= goal:
            count += 1
    return count

steps = [4500, 9200, 7800, 12000, 6100, 10400, 8000]
goal = 8000

print("Total:", total_steps(steps))
print("Average:", average_steps(steps))
print("Best day:", max(steps))
print("First goal day:", first_goal_day(steps, goal))
print("Goal days:", count_goal_days(steps, goal))

for i in range(len(steps)):
    if steps[i] >= goal:
        print("Day", i, "hit the goal")
```

```python
trainees = [
    ("Antman", 71),
    ("Hulk", 92),
    ("Stark", 95),
    ("Wanda", 83),
    ("Barnes", 76),
    ("Parker", 80),
]

def get_successful_trainees(records):
    return [name for name, score in records if score >= 80]


def build_rating_dict(records):
    return {
        name: "Great" if score >= 90 \
        else "Good" if score >= 80 else "Redo"
        for name, score in records
    }

def sort_by_score(records):
    return sorted(records, key=lambda x: x[1], reverse=True)


def get_top_scorers(records, n):
    ranked = sort_by_score(records)
    return [name for name, score in ranked[:n]]


passing = get_successful_trainees(trainees)
grades = build_rating_dict(trainees)
top_students = get_top_scorers(trainees, 2)

print(passing)
# _________________________
# <1m> correct strings; 0.5m if written without ''
# <0.5m> correct order
# <0.5m> in a list
print(top_students)
# _________________________
# <1m> 'Stark', 'Hulk' (nothing less or more)
# <1m> correct order
```

If there is no runtime error, state what would be displayed on the console from `print(build_correct_dict(trainees))`.

```python
def build_correct_dict(records):
    holder = {}
    for name, score in records:
        if 90 <= score <= 100:
            holder[name] = "Great"
            return holder
        elif score < 90:
            holder[name] = "Good"
            return holder
        else:
            holder[name] = "Redo"
            return holder

print(build_correct_dict(trainees))
```
