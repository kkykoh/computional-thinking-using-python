```python
def func(a, b=[]):
    b.insert(0, a)
    return b

print(func(5))
print(func(9))
```

```python
def func(a, b={}):
    b.setdefault(a, []).append(a)
    return b

print(func(1))
print(func(1))
```

```python
import re

pattern = r"(\w+)\s+(\w+)"
text = "John Smith"
result = re.sub(pattern, r"\2, \1", text)
print(result)  # "Smith, John"
```

```python
import re

pattern = r"(\d+)\*"
text = "10*20*30*"

print(re.findall(pattern, text)) 
```

```python
import re

pattern = '(\w+):(\d+)'
text = "a:1 b:2 c:3"

print(re.findall(pattern, text)) 
```

```python
import re

pattern = '(([6-9]\d+)\2)'
text = "167676+"

print(re.findall(pattern, text)) 
```
