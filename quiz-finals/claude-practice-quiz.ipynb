```python
def func(a, b=[]):
    b.insert(0, a)
    return b

print(func(5))
print(func(9))
```

```python
def func(a, b={}):
    b.setdefault(a, []).append(a)
    return b

print(func(1))
print(func(1))
```

```python
import re

pattern = r'src="([^"]*)"'
text = '<img src="pic.png">'

m = (re.search(pattern, text)) 

print(m)
print(m.groups())
print(m.group(0))
print(m.group(1))
```

```python
import re

pattern = r'((\d+)\d\2)'
text = '1234123, 11311, 123, 54345'

m = (re.finditer(pattern, text)) 

for string in m:
    # print(string)
    print(string.group(1,2))
```
