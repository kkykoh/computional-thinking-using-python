```python
def func(a, b=[]):
    b.insert(0, a)
    return b

print(func(5))
print(func(9))
```

```python
def func(a, b={}):
    b.setdefault(a, []).append(a)
    return b

print(func(1))
print(func(1))
```

```python
import re

pattern = r"(\w+)\s+(\w+)"
text = "John Smith"
result = re.sub(pattern, r"\2, \1", text)
print(result)  # "Smith, John"
```

```python
import re

pattern = r'src="([^"]*)"'
text = '<img src="pic.png">'

m = (re.search(pattern, text)) 

print(m)
print(m.groups())
print(m.group(0))
print(m.group(1))
```

```python
import re

pattern = r'alt="([^"]*)"'
text = '<img alt="a cat sitting">'

m = (re.search(pattern, text)) 
n = (re.findall(pattern, text)) 

print(m)
print(m.groups())
print(m.group(0))
print(m.group(1))

print(n)
```

```python
import re

pattern = r'(\w+)=(\d+)'
text = 'x=1, y=2'

m = (re.search(pattern, text)) 

print(m)
print(m.groups())
print(m.group(0))
print(m.group(1))

n = re.findall(pattern, text)
print(n)


p = re.finditer(pattern, text)
print(p)
```

```python
import re

pattern = r'((\d+)\d\2)'
text = '1234123, 11311, 123, 54345'

m = (re.finditer(pattern, text)) 

for string in m:
    # print(string)
    print(string.group(1,2))
```
