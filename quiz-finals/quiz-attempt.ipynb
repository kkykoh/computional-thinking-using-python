```python
data = [7, 12, 5, 18, 11]

def mix(a, b):
    if a > b:
        return a - b
    else:
        return b - a + 1

def scan(items, x):
    for i in range(len(items)):
        if items[i] % x == 0:
            return i
    return -1

def crunch(items):
    result = 0
    for n in items:
        if n % 2 == 0:
            result = result + n // 2
        else:
            result = result - 1
    return result

print(mix(4, 9))
print(scan(data, 6))
print(crunch(data))
print(mix(crunch(data), scan(data, 2)))

final = []
for i in range(len(data)):
    if data[i] % 9 == 0:
        final.append((i, data[i]))

print(final)
```

The following code takes in `trainees`, as a list of tuples, and does some operations to it.


```python
trainees = [
    ("Antman", 71),
    ("Hulk", 92),
    ("Stark", 95),
    ("Wanda", 83),
    ("Barnes", 76),
    ("Parker", 80),
]

def get_successful_trainees(records):
    return [name for name, score in records if score >= 80]

def build_rating_dict(records):
    return {
        name: "Great" if score >= 90 \
        else "Good" if score >= 80 else "Redo"
        for name, score in records
    }


def sort_by_score(records):
    return sorted(records, key=lambda x: x[1], reverse=True)


def get_top_scorers(records, n):
    ranked = sort_by_score(records)
    return [name for name, score in ranked[:n]]


passing = get_successful_trainees(trainees)
grades = build_rating_dict(trainees)
top_students = get_top_scorers(trainees, 2)

print(passing)
print(top_students)
```
