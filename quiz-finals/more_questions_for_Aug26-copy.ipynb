```python
def mystery(num):
    total = 0
    for j in range(1, num):
        if j%3 == 0:
            return total
        else:
            total += j
    return total

mystery(3) + mystery(4)

# A - 6 <--
# B - 9
# C - 12
# D - 16
```

```python
class Stock:
    def __init__(self, ticker, price):
        self.ticker = ticker
        self.price = price    #### (1m)
        #_________________________
    

class Portfolio:
    def __init__(self):
        self.equity = {}
    
    def add_stock(self, stock, num_shares):
        if stock in self.equity:  #### (1m)
        #if ___________________________:
            self.equity[stock] += num_shares
        else:
            self.equity[stock] = num_shares
    
    def total_valuation(self):
        return sum([stock.price * num_shares for stock, num_shares in self.equity.items()])   #### (2m)
        #return sum([stock.price * num_shares _______________________ in _____________________])

PeterPortfolio = Portfolio()
# Add 200 shares of 'APPL' at $179.58
PeterPortfolio.add_stock(Stock('APPL', 179.58), 200)
# Add 300 shares of 'AMZN' at $131.03
PeterPortfolio.add_stock(Stock('AMZN', 131.02), 300)   #### (1m)
#___________________________________________________

assert PeterPortfolio.total_valuation() == 75222
```
