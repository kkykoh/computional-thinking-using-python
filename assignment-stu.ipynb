# Question 2

**Objective**: a class that manages a collection of songs.

Write a class Playlist:
- \_\_init\_\_(self, name) — stores the playlist's name and starts with no songs.
- add_song(self, title, duration) — adds a song (duration is a positive integer, in seconds). If a song with the same title is already in the playlist, nothing is added and the method returns False; otherwise the song is added and it returns True.
- song_count(self) — returns the number of songs in the playlist.
- total_duration(self) — returns the sum of all durations (0 for an empty playlist).
- longest_song(self) — returns the title of the song with the longest duration. If two songs tie, return the one that was added first. Returns None for an empty playlist.


```python
class Playlist:
    ### INSERT YOUR CODE FOR Q2 FOR EACH FUNCTION BELOW ###

    def __init__(self, name):
        self.name = name
        self.songs = []
        # pass

    def add_song(self, title, duration):
        for song_title, song_duration in self.songs:
            if song_title == title:
                return False
        
        self.songs.append((title, duration))
        return True
        # pass

    def song_count(self):
        return len(self.songs)
        # pass

    def total_duration(self):
        return sum(duration for name, duration in self.songs)
        # pass

    def longest_song(self):
        if len(self.songs) == 0:
            return None

        return max(self.songs, key = lambda song: song[1])[0]
        # pass

    ### END OF YOUR CODE FOR Q2 ###


# GIVEN — do not modify. The grader calls q2 with a list of
# (title, duration) pairs and checks the returned tuple:
# (add_song, song_count, total_duration, longest_song)
def q2(songs: list) -> tuple:
    p = Playlist("test")
    adds = []
    for title, duration in songs:
        adds.append(p.add_song(title, duration))
    return (adds, p.song_count(), p.total_duration(), p.longest_song())
```

```python
# sample test cases for Q2
print(q2(([("Shape of You", 240), ("Havana", 217)]))) # expected output: ([True, True], 2, 457, 'Shape of You')
print(q2([("Lofi Rain", 180), ("Lofi Rain", 200)])) # expected output: ([True, False], 1, 180, 'Lofi Rain')
```

# Question 3

**Objective**: a few class(es) that manage ticket selling. 

Write:
- A class Ticket: 
    - \_\_init\_\_(self, holder, base_price) — stores the holder's name and the base price (a positive integer).
    - price(self) — returns the amount this ticket holder pays. For a regular Ticket, this is just the base price.
    - \_\_str\_\_(self) — returns holder pays $X, where X is what that ticket holder actually pays
- A class ChildTicket that inherits from Ticket: children pay half the base price, rounded down to a whole dollar.
- A class SeniorTicket that inherits from Ticket: seniors get $12 off the base price, but the price never goes below $0.
- A standalone function total_revenue(tickets) that takes a list containing any mix of the three ticket types and returns the total amount paid.


```python
### INSERT YOUR CODE FOR Q3 IN THE SPACE BELOW ###
    
class Ticket:
    def __init__(self, holder, base_price):
        self.holder = holder
        self.base_price = base_price
        self.tickets = []
        # pass

    def price(self):
        return self.base_price
        # pass

    def __str__(self):
        return f"{self.holder} pays ${str(self.price())}"
        # pass


class ChildTicket(Ticket):
    def price(self):
        return int(self.base_price/2)
        # pass


class SeniorTicket(Ticket):
    def price(self):
        if self.base_price >=12:
            senior_price = self.base_price - 12
        elif self.price <12:
            senior_price = 0

        return senior_price
        # pass


def total_revenue(tickets):
    return sum(ticket.price() for ticket in tickets)
    # pass

### END OF YOUR CODE FOR Q3 ###

# GIVEN — do not modify. The grader calls q3 with a list of
# (kind, holder, base_price) triples, builds the tickets, and checks
# the returned tuple: (list of prices, list of str(ticket), total_revenue)
def q3(specs: list) -> tuple:
    kinds = {"regular": Ticket, "child": ChildTicket, "senior": SeniorTicket}
    tickets = [kinds[kind](holder, price) for kind, holder, price in specs]
    return ([t.price() for t in tickets], [str(t) for t in tickets], total_revenue(tickets))
```

```python
# sample test cases for Q3
print(q3([("regular", "Ana", 30)])) # expected output: ([30], ['Ana pays $30'], 30)
print(q3([("child", "Ben", 25)])) # expected output: ([12], ['Ben pays $12'], 12)
```

```python
# GIVEN — do not modify. The grader calls q4 with a dictionary of
# dictionaries, where each has keys "year", "runtime", "budget", and "revenue".
# Return a list of titles (as strings) whose revenue is at least 
# 5 times their production budget.

def q4(movies: dict) -> list:
    ### INSERT YOUR CODE FOR Q4 IN THE SPACE BELOW ###
    high_performers = []

    for name, movie in movies.items():
        if (movie["budget"] * 5 < movie["revenue"]):
            high_performers.append(name)
        

    return high_performers
    
    ### END OF YOUR CODE FOR Q4 ###
    # pass
```
