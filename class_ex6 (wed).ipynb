We define a fraction, $\frac{x}{y}$, such that the gratest common divisor (GCD) between $x$ and $y$ is $1$.

For two numbers $p$ and $q$, the GCD is defined as the largest positive number $a$ such that $|p|\;mod\;a = |q|\;mod\;a = 0$. For example, the GCD of $8$ and $6$ is $2$.

In the event that the GCD of $x$ and $y$ is greater than $1$, we can simplify the fraction by dividing both $x$ and $y$ by this GCD. For example, to simplify $x=6$ and $y=8$, we can divide by it's GCD ($2$) to get $\frac{3}{4}$.

Write a class ```Fraction``` that takes in a positive numerator and denominator and contains a method ```simplified``` that prints the simplified fraction.

````python
class Fraction:

    def __init__(self, numerator, denominator):
        ### insert code here ###
        self.numerator = numerator
        self.denominator = denominator
        self.simplified()

    def simplified(self):
        ### insert code here ###
        gcd = min(self.numerator, self.denominator)
        while gcd > 1:
            if self.numerator % gcd == 0 and self.denominator % gcd == 0:
                self.numerator //= gcd # what if we use /= instead?
                self.denominator //= gcd
                break
            gcd -= 1

    def __str__(self):
        ### insert code here ###
        return "The fraction is {}/{}".format(self.numerator, self.denominator)

# for testing purposes
print(Fraction(2, 3))
print(Fraction(6, 4))
print(Fraction(8, 6))
````

The code below defines a *Professor* class.
* Each instance of this class is associated to a professor.
* The instance records the active courses that the professor is currently teaching.
* A professor can teach multiple sections of the same course.
* The workload of a professor is the sum of the complexity of all sections she is currently teaching across all different courses.

````python
class Professor:
    def __init__(self):
        '''Create a new instance, with no active courses.'''

    def add_section(self, course):
        '''Add a new course to this professor.'''

    def deactivate_course(self, course):
        '''Remove all teaching sections of the same course from this professor'''

    def load(self):
        '''Calculate the current load for all courses.'''

    def __str__(self):
        '''Return a string with the current load of the professor'''
    
    def __lt__(self, other):
         return self.load() < other.load()
````

````python
class Professor:
    def __init__(self):
        '''Create a new instance, with no active courses.'''
        self.sections = {} # use a dictionary to record the list of courses he is teaching and how many section he is teaching

    def add_section(self, course):
        '''Add a new course to this professor.'''
        # Add this course to the 'courses' dictionary
        # if he is already teaching this course, add one more section
        if course in self.sections:
            self.sections[course] += 1
        else:
            self.sections[course] = 1

    def deactivate_course(self, course):
        '''Remove all teaching sections of the same course from this professor'''
        # Remove the course from the dictionary
        if course in self.sections:
            del self.sections[course]

    def load(self):
        '''Calculate the current load for all courses.'''
        total_workload = 0.0
        # Add up the load of all sections
        for c in self.sections:
            total_workload += self.sections[c] * c.complexity
        return total_workload

    def __str__(self):
        '''Return a string with the current load of the professor'''
        return '{:.2f}'.format(self.load())
    
    def __lt__(self, other):
         return self.load() < other.load()
````

````python
class LoadBalancing:
    def __init__(self, courses = None):
        '''Initialize with one professor'''
        self.courses = courses if courses else {}
        self.profs = [Professor()]
        self.taught_by = {k:[] for k in courses}

    def add_course(self, course_id):
        '''Add a new course into available courses'''
        if course_id not in self.courses:
            self.courses[course_id] = Course(course_id)
            self.taught_by[course_id] = []

    def add_section(self, course_id):
        '''Select a professor with the lowest load and add a section of that course to him/ her.
           If it's a new course, add the course first'''
        if course_id not in self.courses:
            self.add_course(course_id)
        lowest_load = min(self.profs)
        
        # Add the course to the dictionary with the selected professor
        # Add the course to the professor
        lowest_load.add_section(self.courses[course_id])
        if lowest_load not in self.taught_by[course_id]:
            self.taught_by[course_id].append(lowest_load)

    def deactivate_course(self, course_id):
        '''deactivate the course from all teaching professors.'''
        # Find out professors who is teaching this course
        # deactivate the course for each professor
        # Remove the course from the load balancer
        for p in self.taught_by[course_id]:
            p.deactivate_course(self.courses[course_id])
        del self.courses[course_id]
        del self.taught_by[course_id]

    def avg_load(self):
        '''Calculate the average load of all professors'''
        # Sum the load of each course and divide by the number of professors
        loads = [p.load() for p in self.profs]
        return sum(loads) / len(loads)

    def __str__(self):
        '''Return a string with the load for each professor.'''
        loads = [str(p) for p in self.profs]
        return '[{}]'.format(','.join(loads))
````

We hired a new professor manually. But we want this to happen automatically when the average load is more than 50%. To make this possible, we add a method **ensure_availability** and call it from the method **add_course** after a new course has been added. You can test it with the following code:

````python
class LoadBalancing:
    def __init__(self, courses = None):
        '''Initialize with one professor'''
        self.courses = courses if courses else {}
        self.profs = [Professor()]
        self.taught_by = {k:[] for k in courses}

    def add_course(self, course_id):
        '''Add a new course into available courses'''
        if course_id not in self.courses:
            self.courses[course_id] = Course(course_id)
            self.taught_by[course_id] = []

    def add_section(self, course_id):
        '''Select a professor with the lowest load and add a section of that course to him/ her.
           If it's a new course, add the course first'''
        if course_id not in self.courses:
            self.add_course(course_id)
        lowest_load = min(self.profs)
        
        # Add the course to the dictionary with the selected professor
        # Add the course to the professor
        lowest_load.add_section(self.courses[course_id])
        if lowest_load not in self.taught_by[course_id]:
            self.taught_by[course_id].append(lowest_load)
        self.ensure_availability()

    def deactivate_course(self, course_id):
        '''deactivate the course from all teaching professors.'''
        # Find out professors who is teaching this course
        # deactivate the course for each professor
        # Remove the course from the load balancer
        for p in self.taught_by[course_id]:
            p.deactivate_course(self.courses[course_id])
        del self.courses[course_id]
        del self.taught_by[course_id]

    def avg_load(self):
        '''Calculate the average load of all professors'''
        # Sum the load of each course and divide by the number of professors
        loads = [p.load() for p in self.profs]
        return sum(loads) / len(loads)

    def ensure_availability(self):
        '''If the average load is higher than 50, hire a new professor'''
        if self.avg_load() > 50:
            self.profs.append(Professor())

    def __str__(self):
        '''Return a string with the load for each professor.'''
        loads = [str(p) for p in self.profs]
        return '[{}]'.format(','.join(loads))
````
