```python
class BankAccount:
    def __init__(self, balance = 0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

class Saving(BankAccount):
    def __init__(self, balance = 0, interest_rate = 0):
        super().__init__(balance)
        self.interest_rate = interest_rate

    def deposit(self, *amount):
        am_after_interest = sum([am + am * self.interest_rate for am in amount])
        super().deposit(am_after_interest)


s = Saving(1000, 0.03)
s.deposit(500, 100, 400)
print(s.balance)
```

```python
import re 
html = '<a href="https://example.com">Visit our website</a>' 
href_pattern = r'href="([^"]*)"'
print(re.search(href_pattern, html).group())
print(re.search(href_pattern, html).group(0))
print(re.search(href_pattern, html).group(1))
print(re.search(href_pattern, html).groups())
```
