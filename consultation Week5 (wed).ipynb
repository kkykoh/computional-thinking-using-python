```python
i, upper_bound, summ = 1, 9, 0
while i <= upper_bound:
    summ += i
    i += 1
    upper_bound -= 1

"""
summ,   i,  upper_bound 
0       1,      9,       
1       2       9-1=8
1+2=3   3       8-1=7
    6   4           6
   10   5           5
   15
"""

print(summ)
```

```python
switch = 1
## flip between 0 and 1

switch = 1 - switch   ## 1 - 0 = 1
print("(i)", switch)

switch = 1 - switch   ## 1 - 1 = 0
print("(ii)", switch)
```

```python
c1, c2, s = 'x', 'y', 'ooxooyoyooxooxooyooox'
flag, pos, count = 0, 0, 0
while pos != -1:
    if flag:
        pos = s.find(c1, pos)
    else:
        pos = s.find(c2, pos)
    if pos == -1:
        continue
    flag = 1 - flag
    pos += 1
    count += 1
print(count)
```

```python
mylist = [1, 2, 9, 9, 3, 4, 9]
c = 9
i = 0   # index to start at 0

while i < len(mylist):
    print("mylist", mylist)
    print("i", i, "mylist[i]", mylist[i])
    if mylist[i] == c:
        mylist.pop(i)
    i += 1

print('Ending: ', mylist) 
```

```python
a, b = [1, 2, 3], [4, 1, 1, 9, 2, 2, 1, 2, 5, 3, 3, 8, 2, 2, 6, 1]
#                     1              1                          1
#                              2        2              2
#                                             3  
for c in a:
    i = 0
    while i < len(b):
        if b[i] == c:
            b.pop(i)
        i += 1

print(len(b))
```
