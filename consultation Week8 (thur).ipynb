```python
class A:
    def __init__(self):
        self.a = 1
        self.fun1()
    
    def fun1(self):
        b = 2

    def fun2(self):
        self.c = 3
    
    def fun3(self):
        self.d = 4

a = A()
a.fun2()
```

```python
[x for x in a.__dir__() if not x.startswith('_')]
```

```python
class Z:
    def __init__(self):
        self.a = 1
        self.fun1()
    
    def fun1(self):
        b = 2
        self.fun2()

    def fun2(self):
        self.c = 3
        self.fun3()

    def fun3(self):
        self.d = 4

z = Z()
```

```python
[x for x in z.__dir__() if not x.startswith('__')]
```

```python
class Parent:
    def __init__(self, a):
        self.a = a

class Child(Parent):
    def __init__(self, a, b):
        super().__init__(a)
        self.b = b

print(Child(1, 2).a)
```

```python
print(Child(56, 78).a)
```
