```python
a, b, c = 10, 100, 1000

def operation(a, b, c):
    return 2*a + 3*c

operation(b, c, a)

# Step 1
# operation(b, c, a) is operation(100, 1000, 10)
# a, b, c = 100, 1000, 10

# Step 2
# 2*a + 3*c = 2*(100)+ 3*(10) = 230

# Step 3
# Replace with 230
```
