```python
text = "tan_lee_chan@gmail.com"
# text = "RTX 5090 0905"

p = re.compile(r'[\w]*')
m = p.findall(text)
print(m)
```

```python
holder = ["rush_hour_iii", "mission", "impossible", "jamesbond"]

for text in holder:
    p = re.compile(r'[i-n]{2,}')   # [ijklmn]
    m = p.findall(text)
    print(m)

# [a] rush_hour_iii
# [b] mission
# [c] impossible
# [d] jamesbond
```

```python
text = "Let's look at more examples dealing with character class"

p = re.compile(r'[a-c-g-i]{2}')  # r'[abcghi-][abcghi-]'
m = p.findall(text)
print(m)
```

```python
text = "hello goodbye"
p = re.compile(r'((\w){3})')
m = p.findall(text)
print(m)

# (a) [('hel', 'l'), ('goo', 'o'), ('dby', 'y')]
# (b) [('hel', 'l'), ('goo', 'o')]
# (c) ['hel', 'goo', 'dby']
# (d) ['hel', 'goo']
```

```python
p = re.compile(r'((\w){3,})')
m = p.findall('abcd@efg hijklmn pqr+stu')
print(m)
```

```python
p = re.compile(r'look|at')
m = p.findall('looker cat')
print(m)
```
