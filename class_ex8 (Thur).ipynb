```python
def is_sorted(arr, cmp_func = None):
    if cmp_func is None:
        cmp_func = lambda x, y: x <= y
    
    for i in range(len(arr)-1):
        if not cmp_func(arr[i], arr[i+1]):
            return False
    return True
```

```python
import random
random.seed('sort')
orig_array = [random.randint(1, 1000000) for _ in range(1000)]
```

### Insertion sort

```python
def insertion_sort(A):
    for i in range(1, len(A)):
        k = A[i]
        j = i - 1
        while j >= 0 and A[j] > k:
            A[j+1] = A[j]
            j -= 1
        A[j+1] = k
```

```python
insertion_sort([1, 2, 3, 4, 5, 88, 65, 70])
```

```python
def bubble_sort(A):
    swapped = True
    cnt = 0
    while swapped:
        swapped = False
        for i in range(len(A)-1):
            if A[i] > A[i+1]:
                A[i], A[i+1] = A[i+1], A[i]
                swapped = True
                cnt += 1
```

```python
insertion_sort([1, 2, 3, 4, 5, 88, 65, 70])
```

```python
A1 = orig_array.copy()
insertion_sort(A1)
print('This array is', 'well' if is_sorted(A1) else 'not', 'sorted.')
```

### Extension of radix sort
When processing 1000 customers, priorities are giving to customers who are older in age (primary), and arrive earlier in the system (secondary). We may use radix sort to sort by arrival time, then by age.
Note on the order: sort by the less significant attribute, then by the more significant attribute.

```python
customers = list(zip(random.choices(range(50, 80), k=1000), random.choices(['09:%02d' % i for i in range(60)], k=1000)))
```

```python
customers
```

```python
def radix_sort_by(s, i, cmp_func = None, reverse = False):
    n, d, res = len(s), {}, []
    for j in range(n):
        if s[j][i] in d:
            d[s[j][i]].append(s[j])
        else:
            d[s[j][i]] = [s[j]]
    for k in sorted(d.keys(), key = cmp_func, reverse = reverse):
        res.extend(d[k])
    return res
```

```python
sorted_by_arr_time = radix_sort_by(customers, 1)
```

```python
sorted_by_age_then_arr_time = radix_sort_by(sorted_by_arr_time, 0, reverse = True)
```

```python
print('This array is', 'well' if is_sorted(sorted_by_age_then_arr_time, cmp_func = lambda x, y: x[0] > y[0] or (x[0] == y[0] and x[1] <= y[1])) else 'not', 'sorted.')
```

### Exercise 1: Binary Search

```python
# solution in week 5
def BinarySearch(ls, target):
    min_idx, max_idx = 0, len(ls) - 1
    while min_idx <= max_idx:
        mid_idx = (min_idx + max_idx) // 2
        if ls[mid_idx] == target:
            return mid_idx
        elif ls[mid_idx] < target:
            min_idx = mid_idx + 1
        else:
            max_idx = mid_idx - 1
        print(min_idx, max_idx)
    return -1            
```

```python
# use bitwise operators
def BinarySearch_bit(lst, target):
    min_idx, max_idx = 0, len(lst) - 1
    while min_idx <= max_idx:
        mid_idx = (min_idx + max_idx) >> 1
        if lst[mid_idx] == target:
            return mid_idx
        elif lst[mid_idx] < target:
            min_idx = mid_idx + 1
        else:
            max_idx = mid_idx - 1
        print(min_idx, max_idx)
    return -1     
```

```python
lst = [3, 11, 21, 34, 46, 65, 66, 97, 98, 99, 100, 234, 456, 555, 564, 598, 666]
target = 65
BinarySearch_bit(lst, target)
```

### Exercise 2: Hire chefs

```python
chefs = {
    'John': ['Apple Pie', 'Bak Kut Teh', 'Egg Prata'], 
    'Zach': ['Chilli Crab'],
    'Amy': ['Bak Kut Teh', 'Duck rice', 'Chilli Crab'], 
    'Bobby': ['Duck rice', 'Fish Soup']
}
```

```python

def naive_maxDishes(chefs):
    chef1 = chef2 = ''
    res = 0
    for i in chefs:
        for j in chefs:
            ## example
            ## i = 'John'
            ## chefs[i] = ['Apple Pie', 'Bak Kut Teh', 'Egg Prata'], 
            ## j = 'Amy'
            ## chefs[j] = ['Bak Kut Teh', 'Duck rice', 'Chilli Crab'], 
            common = False
            for d in chefs[j]:  # j is the key (chef name), chefs[j] will be the list of dishes
                if d in chefs[i]:
                    common = True
                    break

            if (not common) and (len(chefs[i]) * len(chefs[j]) > res):
                res = len(chefs[i]) * len(chefs[j])
                chef1 = i
                chef2 = j

    return chef1, chef2, res
```

```python
naive_maxDishes(chefs)
```

```python
def maxDishes(chefs):
    n = len(chefs)
    res = 0
    chef1 = chef2 = ''
    
    masks = {}
    for chef in chefs:
        masks[chef] = 0
        for dish in chefs[chef]:
            masks[chef] |= 1 << ord(dish[0]) - ord('A')
    ####
    for i in chefs:
        for j in chefs:
            if not (masks[i] & masks[j]) and (len(chefs[i]) * len(chefs[j]) > res):
                res = len(chefs[i]) * len(chefs[j])
                chef1 = i
                chef2 = j
    
    return chef1, chef2, res
```

```python
import mmh3
from bitarray import bitarray

class BloomFilter:
    def __init__(self, size, num_hash):
        self.size = size
        self.num_hash = num_hash
        self.bit_array = bitarray(size)
        self.bit_array.setall(0)

    def add(self, username):
        # Hash the username and set the corresponding bits in the bit array
        for i in range(self.num_hash):
            # Use MurmurHash to generate hash
            hash_value = mmh3.hash(username, i) % self.size
            self.bit_array[hash_value] = 1

    def check(self, username):
        # Check if all bits at the positions are set to 1
        return all(self.bit_array[mmh3.hash(username, i) % self.size] == 1 for i in range(self.num_hash))
```

```python
system = BloomFilter(8, 2)

# Adding usernames
system.add("Alice")
system.add("Bob")

# Checking for usernames
to_be_checked = ['Alice', 'Bob', 'Cindy']
for name in to_be_checked:
    if system.check(name):
        print('Username ' + name + ' already exists') # might be false positive
    else:
        print('Username ' + name + ' can be used.')
```
