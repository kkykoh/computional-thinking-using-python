```python
def is_leap_year(year:int):
    if year % 4 != 0:          # checking for numbers that are not multiples of 4
        return False
    elif year % 100 != 0:
        return True
    elif year % 400 != 0:  # years like 1800, 1900, 2100 are not leap years although they are divisible by 4
        return False
    else:                  # years like 1600, 2000, 2400 are.
        return True
```

```python
def last_day_of_month(year:int, month:int):
    if month <= 0 or month > 12:
        print('not a valid month')
        return -1
    if month == 2:
        if is_leap_year(year):
            return 29
        else:
            return 28
    elif (month <= 7) == (month % 2):
        return 31
    else:
        return 30
```

```python
def info(name, **kwargs):
    hobby = []
    for k, v in kwargs.items():
        hobby.append(k + '-' + v)
    print(name + "'s hobbies: " + ', '.join(hobby))

info('Mike', first = 'hiking', second = 'reading')
# Mike's hobbies: first-hiking, second-reading
```

```python
def f2():
    s2='456'
    print('before ', s2)
    s2 += '0000'
    print('after ', s2)
    print(s2)
    print("warning: aft this, s2 will be entirely forgotten")

s2='123' # set as a global
f2() # a local overwrite --> in this function, use s2=456
print(s2)   # we are not in the function anymore so it is back to global var value --> 123
```
