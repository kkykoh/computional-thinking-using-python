## Regular Expressions

````python
import re
````

There are 3 main steps when working with regular expressions:

1. Coming up with a pattern

2. Compiling the pattern using ```re.compile()```

3. Using the pattern to search for a text using either ```match```, ```search```, ```findall```, or ```finditer```.

````python
pat = re.compile(r'abc')
print(pat)
print(type(pat))
````

````python
mat_abc1 = pat.match('ABC,ABc,AbC,abc')
mat_abc2 = pat.match('abc,ABc,AbC,abc')
print(mat_abc1)
print(mat_abc2)
````

````python
sear_abc1 = pat.search('ABC,ABc,AbC,abc')
sear_abc2 = pat.search('abc,ABc,AbC,abc')
print(sear_abc1)
print(sear_abc2)
print(type(sear_abc1))
````

````python
find_abc1 = pat.findall('ABC,ABc,AbC,abc')
find_abc2 = pat.findall('abc,ABc,AbC,abc')
print(find_abc1)
print(find_abc2)
````

````python
finditer_abc = pat.finditer('abc,ABc,abc,abc')

print(finditer_abc)

for m in finditer_abc:
    print(m)
    
````

### Metacharacters

#### Metacharacters that match one character

##### Dot (.)

````python
# ex1
p = re.compile(r'.')
m = p.findall('a1\nA#')
print(m)
````

````python
# ex2
p = re.compile(r'.at')
m = p.findall('cat bat sat cap')
print(m)
````

##### Character Class ([])

````python
# ex1
p = re.compile(r'[aA]')
m = p.findall('a1\nA#')
print(m)
````

````python
# ex2
p = re.compile(r'[abcABC]')
m = p.findall('abcABC')
print(m)
````

````python
# ex3: [a-zA-Z0-9]
p = re.compile(r'[a-z0-9]')
m = p.findall('d0A3z6P')
print(m)

# ex4: 
p = re.compile(r'[C-Eb-c2-4]')
m = p.findall('ABCDEF abcd 012345')
print(m)

# ex5:
p = re.compile(r'[-a-e]') # or [a-e-]
m = p.findall('e-a-s-y, easy')
print(m)
````

````python
# ex6: [^]
p = re.compile(r'[^0-9a-z]')
m = p.findall('1 2 3 Go')
print(m)

# ex7:
p = re.compile(r'[0-9^a-z]')
m = p.findall('1 2 3 ^Go')
print(m)
````

##### Digits (\d) and \D

````python
# ex1
p = re.compile(r'\d')
m = p.findall('a1\nA#')
print("\\d matches: ", m)
````

````python
# ex2
p = re.compile(r'\D')
m = p.findall('a1\nA#')
print("\\D matches: ", m)
````

##### Word Characters (\w) and \W

````python
# ex1
p = re.compile(r'\w')
m = p.findall('_#a!E$4-')
print("\\w matches: ", m)
````

````python
# ex2
p = re.compile(r'\W')
m = p.findall('_#a!E$4-')
print("\\W matches: ", m)
````

##### Whitespaces (\s) and \S

````python
# ex1
text = 'Name\tISSS610\tISSS666\nJoe Jones\tA\tA\n'
print(text)

p = re.compile(r'\s')
m = p.findall(text)
print("\\s matches: ", m)
````

````python
# ex2
p = re.compile(r'\S')
m = p.findall(text)

print("\\S matches: ", m)
````

#### Escaping Metacharacters

##### Backslash (\\) as Escaping Character

````python
# ex1
p1 = re.compile(r'.')
p2 = re.compile(r'\.')
m1 = p1.search('smu.edu.sg')
m2 = p2.search('smu.edu.sg')
print(m1, m2)
````

````python
# ex2
p = re.compile(r'\d\\d')
m = p.findall('135\\d')
print(m)
````

#### Anchors

##### Caret (^)

````python
# ex1
p = re.compile(r'^ab.')
m = p.findall('abc abd abe abf')
print(m)
````

````python
# ex2
p = re.compile(r'^a[ab]c')
m = p.findall('''aac\nabc''')
print(m)

p = re.compile(r'^a[ab]c', re.M)
m = p.findall('''aac\nabc''')
print(m)
````

##### Dollar ($)

````python
# ex1
p = re.compile(r'ab.$')
m = p.findall('abc abd abe abf')
print(m)
````

````python
# ex2
p = re.compile(r'[ab]c$')
m = p.findall('ac\nbc')
print(m)

p = re.compile(r'[ab]c$', re.M)
m = p.findall('ac\nbc')
print(m)
````

##### Word Boundaries (\b) and \B

````python
# ex1
p = re.compile(r'\b\d\d\b')
m = p.findall('1 2 3 11 12 13 111 112 113')
print(m)
````

````python
# ex2
p = re.compile(r'\b\w\w\b')
m = p.findall('aa,ab;ac(AA)AB AC')
print(m)
````

#### Quantifiers

##### Asterisk (\*), Question Mark (?) and Plus (+)

````python
p = re.compile(r'a[ab]*c')
m = p.findall('a ab ac abc aac aabc aaac ababc')
print(m)
````

````python
p = re.compile(r'a[ab]+c')
m = p.findall('a ab ac abc aac aabc aaac ababc')
print(m)
````

````python
p = re.compile(r'a[ab]?c')
m = p.findall('a ab ac abc aac aabc aaac ababc')
print(m)
````

##### {m} and {m,n}

````python
p = re.compile(r'\d{3}')
m = p.findall('1 2 3 11 12 13 111 112 113')
print(m)
````

````python
p = re.compile(r'\d{2,3}')
m = p.findall('1 2 3 11 12 13 111 112 113')
print(m)
````

#### Grouping Constructs

##### Groups (())

````python
p = re.compile(r'(\w+): (\d+)')
m = p.findall('Course: Grade\nMath: 89\nPhysics: 92\n English: 78')
print(m)
````

##### Using Quantifiers with Groups

````python
p = re.compile(r'(ha)+')
m = p.findall('ha hh aa hahahaha')
print(m)
````

````python
chapters = 'Chapter 12: Numpy\n\
Chapter 13: Pandas\n\
Chapter 14: Data Visualzation'
p = re.compile(r'^Chapter (\d+: .+)', re.M)
m = p.findall(chapters)
print(m)
````

##### Alternation (|)

````python
p = re.compile(r'(\w+)\.(bat|zip|exe)')
m = p.findall('game.exe auto.bat text.zip')
print(m)
````

##### Capturing Groups

````python
p = re.compile(r'(\w+\.\w+)\s(\w+\.\w+)')
m = p.search('game.exe auto.bat text.zip')
print(m)
print(m.groups())
````

````python
pattern = r'(\w+)\W+(\w+)\W+(\w+)\W+(\w)+'
p = re.compile(pattern)
m = p.search('one,,,two:three++++++4')
print(m.group(0))
print(m.group(1))
print(m.group(2, 3, 4))
````

##### Backreference

````python
p = re.compile(r'((\d+)\d\2)')
m = p.finditer('1234123, 11311, 123, 54345')
for string in m:
    print(string.group(1, 2))
````

#### Flags

````python
# re.I
p1 = re.compile(r'abc')
m1 = p1.findall('abc ABC aBC Abc')
p2 = re.compile(r'abc', re.I)
m2 = p2.findall('abc ABC aBC Abc')
````

````python
# re.S
p1 = re.compile(r'.')
m1 = p1.findall('Aa1! \n')
p2 = re.compile(r'.', re.S)
m2 = p2.findall('Aa1! \n')
````
