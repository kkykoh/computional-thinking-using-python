```python
def addition(num1, num2):
    return(num1 + num2)

def subtraction(num1, num2):
    print(num1 - num2)
    # return num1 - num2

def main_func(num1, num2, num3, num4):
    added_num = addition(num1, num2)           # return 2 + 4
    new_added_num = addition(added_num, num3)  # return 6 + 7
    print(subtraction(new_added_num, num4))    # 13 - 10

main_func(2, 4, 7, 10)

# 3      None     3   3
# None   None         3
```

```python
def string_func(*data):
    res = ''
    for word in data:
        if len(word) <= 5:
            new_letters = word[1:4].upper()   # word[1] word[2] word[3]
            if new_letters[0] == new_letters[1]:
                res += new_letters    # append 3 chars to res (initially '')
        else:
            res += word[-2:20]
    return res
                               # OOR    # ns
# print(len(string_func("green", "door", "mountains")))
print(string_func("green", "door", "mountains"))

# 5
```
