```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.titles = []
        self.durations = []

    def add_song(self, title, duration):
        if title in self.titles:      # or an explicit linear-search loop
            return False
        self.titles.append(title)
        self.durations.append(duration)
        return True

    def song_count(self):
        return len(self.titles)

    def total_duration(self):
        total = 0
        for d in self.durations:
            total = total + d
        return total

    def longest_song(self):
        if len(self.titles) == 0:
            return None
        longest_i = 0
        for i in range(1, len(self.titles)):
            if self.durations[i] > self.durations[longest_i]:
                longest_i = i
        return self.titles[longest_i]

    ### END OF YOUR CODE FOR Q2 ###
    pass

# GIVEN — do not modify. The grader calls q2 with a list of
# (title, duration) pairs and checks the returned tuple:
# (add_song results, song_count, total_duration, longest_song)
def q2(songs: list) -> tuple:
    p = Playlist("test")
    adds = []
    for title, duration in songs:
        adds.append(p.add_song(title, duration))
    return (adds, p.song_count(), p.total_duration(), p.longest_song())
```

```python
class Ticket:
    def __init__(self, holder, base_price):
        self.holder = holder
        self.base_price = base_price

    def price(self):
        return self.base_price

    def __str__(self):
        return self.holder + " pays $" + str(self.price())


class ChildTicket(Ticket):
    def price(self):
        return self.base_price // 2


class SeniorTicket(Ticket):
    def price(self):
        discounted = self.base_price - 12
        if discounted < 0:
            return 0
        return discounted


def total_revenue(tickets):
    total = 0
    for t in tickets:
        total = total + t.price()
    return total



# GIVEN — do not modify. The grader calls q3 with a list of
# (kind, holder, base_price) triples, builds the tickets, and checks
# the returned tuple: (list of prices, list of str(ticket), total_revenue)
def q3(specs: list) -> tuple:
    kinds = {"regular": Ticket, "child": ChildTicket, "senior": SeniorTicket}
    tickets = [kinds[kind](holder, price) for kind, holder, price in specs]
    return ([t.price() for t in tickets], [str(t) for t in tickets], total_revenue(tickets))
```

```python
# sample test cases for Q3
print(q3([("regular", "Ana", 30)]))
print(q3([("child", "Ben", 25)]))
```

```python
# GIVEN — do not modify. The grader calls q4 with a dictionary of
# dictionaries, where each has "year", "runtime", "budget", and "revenue".
# Return a list of titles (dictionary keys) whose revenue is at least 
# 5 times their production budget.

def q4(movies: dict) -> list:
    result = [
        title
        for title, info in movies.items()
        if info["revenue"] >= 5 * info["budget"]
    ]
    return result
```

```python
# GIVEN — do not modify. The grader calls q5 with a coffee name and 
# one or more ingredient volumes as keyword arguments. 
# Return a tuple in the form (drink_name, total_volume, contains_milk)
# total_volume is the sum of all ingredient volumes. 
# contains_milk should be True if any ingredient name contains "milk"  
# (e.g. "milk", "oat_milk", "skim_milk"), otherwise False.

def q5(drink_name, **kwargs):
    total_volume = sum(kwargs.values())
    contains_milk = any("milk" in ingredient for ingredient in kwargs.keys())

    return (drink_name, total_volume, contains_milk)
```

```python
class Soldier:
    def __init__(self):
        self.health = 50
        self.attack = 20
        self.defense = 10

class Equipment:
    def __init__(self, bonus):
        self.bonus_health = bonus[0]
        self.bonus_attack = bonus[1]
        self.bonus_defense = bonus[2]

class Hero(Soldier):
    def __init__(self, list_of_equipments=None):
        super().__init__()
        if list_of_equipments:
            for Eqm in list_of_equipments:
                self.health += Eqm.bonus_health
                self.attack += Eqm.bonus_attack
                self.defense += Eqm.bonus_defense


def q6(own, other):
    own_health = own.health - max(0, other.attack - own.defense)
    enemy_health = other.health - max(0, own.attack - other.defense)
    # damage and health cannot be negative. The lowest is 0
    own_health = max(0, own_health)
    enemy_health = max(0, enemy_health)
    # print(own_health, enemy_health)
    
    if own_health == enemy_health:
        return 0   # draw
    else:
        # 1 if player1 ends with a higher health
        # 2 if player2 ends with a higher health
        return 1 if own_health > enemy_health else 2
        
```

```python
# sample test cases for Q6

sword = Equipment([0, 10, 3])
shield = Equipment([5, 0, 9])
spear = Equipment([0, 14, 7])

FighterA = Soldier()
FighterB = Hero([sword])
print(q6(FighterA, FighterB))

FighterC = Hero([sword, shield])  # health=50+5, attack=20+10, defense=10+12
FighterD = Hero([spear])          # health=50+0, attack=20+14, defense=10+7
print(q6(FighterC, FighterD))
```
